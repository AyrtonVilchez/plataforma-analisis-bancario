```python
%sql
-- Creación segura (no da error si el catálogo ya existe) junto con un comentario descriptivo
CREATE CATALOG IF NOT EXISTS banco
COMMENT 'Este catálogo se utiliza para almacenar los datos del proyecto de analisis bancario';
```

```python
%sql
-- Creación segura de esquema
CREATE SCHEMA IF NOT EXISTS banco.banking_analytics
COMMENT 'Esquema para el proyecto de analisis bancario';
```

```python
%sql
-- Creaciónn segura de volumenes
CREATE VOLUME IF NOT EXISTS banco.banking_analytics.banking_files
COMMENT 'Volumen para el proyecto de analisis bancario';
```

```python
%sql
-- Verificación de lo creado
DESCRIBE SCHEMA banco.banking_analytics;

DESCRIBE VOLUME banco.banking_analytics.banking_files;
```
