```python
# ============================================
# CONFIGURACIÓN DEL SCHEMA BRONZE
# ============================================

from pyspark.sql import functions as F
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType
)


BASE_PATH = "/Volumes/banco/banking_analytics/banking_files"

RAW_PATH = f"{BASE_PATH}/raw"

CATALOG = "banco"
SCHEMA = "banking_analytics"

# ============================================
# RUTAS DE LAS FUENTES
# ============================================

BRANCHES_PATH = f"{RAW_PATH}/branches/branches.csv"
CUSTOMERS_PATH = f"{RAW_PATH}/customers/customers.csv"
ACCOUNTS_PATH = f"{RAW_PATH}/accounts/accounts.csv"
TRANSACTIONS_PATH = f"{RAW_PATH}/transactions/transactions.jsonl"

print("Configuración de ingesta completada.")
```

```python
# ============================================
# LECTURA DE ARCHIVO RAW DE AGENCIAS(BRANCHES)
# ============================================

df_branches = (
    spark.read
        .format("csv")
        .option("header", "true")
        .option("inferSchema", "false")
        .load(BRANCHES_PATH)
)

# ============================================
# INCORPORAR METADATOS
# ============================================

df_branches_bronze = (
    df_branches
        .withColumn("_ingestion_timestamp", F.current_timestamp())# columna de log
        .withColumn("_source_file", F.col("_metadata.file_name")) # columna de archivo de origen
        .withColumn("_ingestion_date", F.current_date()) # columna de fecha de carga
)

display(df_branches_bronze.limit(10))
```

```python
# ============================================
# ESCRITURA EN DELTA LAKE
# ============================================

BRONZE_BRANCHES_TABLE = (f"{CATALOG}.{SCHEMA}.bronze_branches")

(
    df_branches_bronze.write
        .format("delta")
        .mode("overwrite")
        .saveAsTable(BRONZE_BRANCHES_TABLE)
)

print("Tabla Bronze Branches creada correctamente.")
```

```python
# Validaciones de la tabla Bronze Branches
df_bronze_branches = spark.table(
    "banco.banking_analytics.bronze_branches"
)

print("Total de registros:", df_bronze_branches.count())

df_bronze_branches.printSchema()

display(df_bronze_branches.limit(10))
```
